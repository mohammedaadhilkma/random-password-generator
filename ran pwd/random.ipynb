```python
import random

print("Random Password Generator")
randomchars="abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789!@#$%^&*?"
noofpasswords = int(input("Enter the number of passwords to generate: "))
passwordlength = int(input("Enter the length of the password: "))

print("\nGenerated Passwords:")
for i in range(noofpasswords):
    password = ''.join(random.sample(randomchars, passwordlength))
    print(password)
```
